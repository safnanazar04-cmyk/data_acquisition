```python
# Part C – Data Transformation 
# Task 6 
# Rename columns wherever necessary so that all datasets have a common column 
# name for merging. 
# Hint: 
# ● Example: Country → country
```

```python
# GDP dataset
excel_data.rename(columns={"Country": "country"}, inplace=True)

# Literacy_rate dataset
literacy.rename(columns={"name": "country"}, inplace=True)
```

```python
# Task 7 
# Convert the required columns into appropriate numeric data types. 
```

```python
df["population"] = pd.to_numeric(df["population"], errors="coerce")

excel_data["GDP"] = pd.to_numeric(excel_data["GDP"], errors="coerce")

data["internet_users"] = pd.to_numeric(data["internet_users"], errors="coerce")

literacy["literacy_rate"] = pd.to_numeric(literacy["literacy_rate"], errors="coerce")
```

```python
# Task 8 
# Merge all four datasets into one DataFrame. 
# Hint: 
# ● Use the common column (country) 
# ● Choose an appropriate join method.
```

```python
merged_df = df.merge(excel_data, on="country", how="inner")
merged_df = merged_df.merge(data, on="country", how="inner")
merged_df = merged_df.merge(literacy, on="country", how="inner")

print(merged_df.head())
```

```python
# Task 9 
# Create the following calculated columns: 
# 1. Internet Penetration (%) 
# 2. GDP Per Capita 
# Hint: 
# Round the calculated values to two decimal places. 
```

```python
merged_df["Internet Penetration (%)"] = (
    (merged_df["internet_users"] / merged_df["population"]) * 100
).round(2)

merged_df["GDP Per Capita"] = (
    merged_df["GDP"] / merged_df["population"]
).round(2)

print(merged_df.head())
```

```python
# Part D – Load Data 
# Task 10 
# Create a SQLite database. Store the final merged dataset into a table named 
# ‘country_statistics’ 
# Hint: 
# ● Use the sqlite3 library. 
# ● Load the DataFrame using pandas.
```

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("country_statistics.db")
```

```python
merged_df.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Task 11 
# Verify that the data has been loaded successfully. 
# Hint: 
# Retrieve the first five records from the SQLite table. 
```

```python
result = pd.read_sql("SELECT * FROM country_statistics;", conn)
print(result.head())
```

```python
# Part F – Data Analysis Using SQL 
# Answer the following questions using SQLite queries. 
# 11 Display the Top 5 countries by GDP. 
# 12 Find the average literacy rate. 
# 13 Display all countries having literacy rate above 90%. 
# 14 Find the Top 5 countries with the highest internet penetration. 
# 15 Calculate the total population. 
# 16 Find the country with the highest GDP per capita.  
# 17 Export the final merged dataset as a CSV file.  
# 18 Sort the dataset in descending order of GDP per Capita and display the Top 10 
# countries.
```

```python
query = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5;
"""
print(pd.read_sql(query, conn))
```

```python
query = """
SELECT AVG([literacy_rate]) AS Average_Literacy_Rate
FROM country_statistics;
"""
print(pd.read_sql(query, conn))
```

```python
query = """
SELECT country, [literacy_rate]
FROM country_statistics
WHERE [literacy_rate] > 90;
"""
print(pd.read_sql(query, conn))
```

```python
query = """
SELECT country, [Internet Penetration (%)]
FROM country_statistics
ORDER BY [Internet Penetration (%)] DESC
LIMIT 5;
"""
print(pd.read_sql(query, conn))
```

```python
query = """
SELECT SUM(population) AS Total_Population
FROM country_statistics;
"""
print(pd.read_sql(query, conn))
```

```python
query = """
SELECT country, [GDP Per Capita]
FROM country_statistics
ORDER BY [GDP Per Capita] DESC
LIMIT 1;
"""
print(pd.read_sql(query, conn))
```
